# Booking site data scraping

```python
import requests
```

```python
from bs4 import BeautifulSoup
```

```python
import pandas as pd
```

```python
url = "https://www.travelwings.com/ng/en/offers/book-cheap-flights.html?utm_source=google&utm_medium=cpc&utm_campaign=Nigeria-Air-INTL-Destinations-21July23&utm_adgroup=149344547697&utm_term=pnr%20flights&creative=666316329553&device=c&matchtype=b&targetid=kwd-298894411575&gad_source=5&gad_campaignid=20380611321&gclid=EAIaIQobChMI3enb56u3jQMVo0n_AR3gAAcCEAAYASAAEgKB2fD_BwE"
```

```python
response = requests.get(url)
```

```python
response = response.content
```

```python
soup = BeautifulSoup(response, 'html.parser')
```

```python
ul = soup.find('ul', class_='al_routes_ul')
```

```python
li = ul.find_all('li', class_='al_routes_li')
```

```python
flights = []
for list in li:
    locations = list.find_all('div', class_='al_or_de')
    origin = locations[0].strong.text.strip()
    destination = locations[1].strong.text.strip()
    prices = list.find('div', class_='al_price').find('strong').text
    #price = prices.attrs['strong']
    flights.append([origin, destination, prices])
```

```python
df = pd.DataFrame(flights, columns=['Origin', 'Destination', 'Price'])
```

```python
df.to_csv('flights_data.csv')
```
